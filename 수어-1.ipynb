```python
import numpy as np

def calculate_angle(v1, v2):
    """벡터 v1과 v2 사이의 각도를 degree로 반환"""
    v1 = v1 / np.linalg.norm(v1)
    v2 = v2 / np.linalg.norm(v2)
    angle = np.arccos(np.clip(np.dot(v1, v2), -1.0, 1.0))  # arccos(dot)
    return np.degrees(angle)
```

```python
import cv2
import mediapipe as mp
import numpy as np
import csv
import os

# Mediapipe Hands 초기화
mp_hands = mp.solutions.hands
hands = mp_hands.Hands(static_image_mode=False,
                       max_num_hands=1,
                       min_detection_confidence=0.5)
mp_drawing = mp.solutions.drawing_utils

# 벡터간 각도 계산 함수
def calculate_angle(v1, v2):
    v1 = v1 / np.linalg.norm(v1)
    v2 = v2 / np.linalg.norm(v2)
    angle = np.arccos(np.clip(np.dot(v1, v2), -1.0, 1.0))
    return np.degrees(angle)

# 벡터 정의 및 각도 추출
def extract_hand_angles(landmarks):
    points = np.array([[lm.x, lm.y, lm.z] for lm in landmarks])

    angles = []

    # 엄지: 0→1→2, 1→2→3, 2→3→4
    for i in [0, 1, 2]:
        v1 = points[i+1] - points[i]
        v2 = points[i+2] - points[i+1]
        angles.append(calculate_angle(v1, v2))

    # 검지~새끼까지: 0→5→6→7→8 ...
    base_ids = [0, 5, 9, 13, 17]
    for base in base_ids[1:]:  # 검지~소지
        for i in range(base, base+2):
            v1 = points[i+1] - points[i]
            v2 = points[i+2] - points[i+1]
            angles.append(calculate_angle(v1, v2))

    return angles  # 총 3(엄지) + 4×2(손가락) = 11개 각도

# 영상 열기
video_path = 'sample.mp4'
cap = cv2.VideoCapture(video_path)

# CSV 파일 생성
csv_filename = 'hand_angles_dataset.csv'
csv_file = open(csv_filename, mode='w', newline='')
csv_writer = csv.writer(csv_file)
csv_writer.writerow([f'angle_{i}' for i in range(11)])  # 헤더

# 프레임별 처리
while True:
    ret, frame = cap.read()
    if not ret:
        break

    image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = hands.process(image_rgb)

    if results.multi_hand_landmarks:
        for hand_landmarks in results.multi_hand_landmarks:
            angles = extract_hand_angles(hand_landmarks.landmark)
            csv_writer.writerow(angles)

cap.release()
csv_file.close()
```

```python
import cv2
import mediapipe as mp
import numpy as np
import csv
import os

# Mediapipe 설정
mp_hands = mp.solutions.hands
hands = mp_hands.Hands(static_image_mode=False,
                       max_num_hands=2,
                       min_detection_confidence=0.5)
mp_drawing = mp.solutions.drawing_utils

# 각도 계산 함수
def calculate_angle(v1, v2):
    v1 = v1 / np.linalg.norm(v1)
    v2 = v2 / np.linalg.norm(v2)
    angle = np.arccos(np.clip(np.dot(v1, v2), -1.0, 1.0))
    return np.degrees(angle)

# 각 손에서 11개 벡터 각도 추출
def extract_hand_angles(landmarks):
    points = np.array([[lm.x, lm.y, lm.z] for lm in landmarks])
    angles = []
    for i in [0, 1, 2]:
        v1 = points[i+1] - points[i]
        v2 = points[i+2] - points[i+1]
        angles.append(calculate_angle(v1, v2))
    for base in [5, 9, 13, 17]:
        for i in range(base, base+2):
            v1 = points[i+1] - points[i]
            v2 = points[i+2] - points[i+1]
            angles.append(calculate_angle(v1, v2))
    return angles

# CSV 파일 생성
csv_path = 'both_hand_angles_dataset.csv'
is_new = not os.path.exists(csv_path)
csv_file = open(csv_path, mode='a', newline='')
csv_writer = csv.writer(csv_file)
if is_new:
    header = [f'L_angle_{i}' for i in range(11)] + [f'R_angle_{i}' for i in range(11)]
    csv_writer.writerow(header)

# 웹캠 열기
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = hands.process(image_rgb)

    # 그리기용 프레임
    image_bgr = cv2.cvtColor(image_rgb, cv2.COLOR_RGB2BGR)

    # 양손 결과 저장용
    left_angles = None
    right_angles = None

    if results.multi_hand_landmarks and results.multi_handedness:
        for hand_landmarks, handedness in zip(results.multi_hand_landmarks, results.multi_handedness):
            label = handedness.classification[0].label  # 'Left' or 'Right'
            angles = extract_hand_angles(hand_landmarks.landmark)
            mp_drawing.draw_landmarks(image_bgr, hand_landmarks, mp_hands.HAND_CONNECTIONS)

            if label == 'Left':
                left_angles = angles
            elif label == 'Right':
                right_angles = angles

        # 양손 모두 인식되었을 때만 CSV에 저장
        if left_angles and right_angles:
            combined = left_angles + right_angles
            csv_writer.writerow(combined)
            print("✅ 저장:", combined)

    cv2.imshow("양손 수어 데이터 수집 - 'q'로 종료", image_bgr)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
csv_file.close()
print("📁 데이터 저장 완료:", csv_path)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Masking

# 1. 데이터 불러오기
df = pd.read_csv('total.csv')  # 통합된 수어 CSV

# 2. 시퀀스 자르기
sequence_length = 30
X_sequences = []
y_sequences = []

for label in df['label'].unique():
    df_label = df[df['label'] == label]
    for i in range(0, len(df_label) - sequence_length + 1, sequence_length):
        seq = df_label.iloc[i:i+sequence_length, :-1].values  # label 제외
        X_sequences.append(seq)
        y_sequences.append(label)

X = np.array(X_sequences)
y = np.array(y_sequences)

# 3. 라벨 인코딩
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# 4. 학습/검증 분할
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, stratify=y_encoded, test_size=0.2, random_state=42)

# 5. 모델 구성
model = Sequential([
    Masking(mask_value=0., input_shape=(sequence_length, X.shape[2])),
    LSTM(64, return_sequences=True),
    LSTM(64),
    Dropout(0.5),
    Dense(64, activation='relu'),
    Dense(len(le.classes_), activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()

# 6. 모델 학습
model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=30, batch_size=32)

# 7. 성능 평가
loss, acc = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {acc:.4f}")
```
